## 2. Les types de données (et le typage dynamique)
En Python, on manipule principalement 4 types de base :
- `str` (*string*) : du texte entre guillemets (`"bonjour"`, `"phishing"`).
- `int` (*integer*) : des nombres entiers (`18`, `-5`, `400`).
- `float` : des nombres à virgule / décimaux (`3.14`, `0.75`).
- `bool` (*boolean*) : les vérités logiques, uniquement `True` (vrai) ou `False` (faux).

En Python, le type est **deviné automatiquement** et peut changer au fil de l'exécution (typage dynamique).

```python
x = 5
print(x, "est de type :", type(x))

x = "maintenant je suis du texte !"
print(x, "est de type :", type(x))
```

## 5. Les conditions (`if` / `elif` / `else`) : prendre des décisions
Un programme de détection doit réagir différemment selon la dangerosité d'un message.
La structure conditionnelle permet d'exécuter du code **seulement si** une condition est vraie.

> ⚠️ **Attention à l'indentation :** En Python, les lignes de code sous un `if` ou un `else` doivent être décalées de **4 espaces** vers la droite. C'est ce décalage qui indique à Python que ces lignes appartiennent au bloc !

```python
texte = "urgent cliquez ici"

if "urgent" in texte:
    print("⚠️ Attention : e-mail suspect détecté !")
else:
    print("✅ Aucun indice d'urgence : message à priori normal.")
```

## 7. Les fonctions : fabriquer nos propres outils réutilisables
Une fonction est une **recette** ou une **machine** :
- Elle a un nom (après le mot-clé `def`).
- Elle reçoit des données en entrée (les **paramètres** entre parenthèses).
- Elle renvoie un résultat avec le mot-clé `return`.

> 💡 **Différence essentielle :**
> - `print(...)` sert uniquement à afficher un texte dans la console pour l'humain.
> - `return ...` renvoie la vraie valeur à votre programme pour qu'on puisse la stocker dans une variable ou faire un test !
